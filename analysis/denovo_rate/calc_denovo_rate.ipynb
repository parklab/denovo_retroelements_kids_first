```python
sf_prefix="/Users/simon/research/projects/xTEA_denovo_rslts/rslts_kids_first/summary_results/denovo_rate/"
sf_in=sf_prefix+"ori_input.txt"
sf_out=sf_prefix+"calculated_CI.txt"
with open(sf_in) as fin_in, open(sf_out,"w") as fout:
    fout.write("Study\tTrios\tRetroelements\tType\tAverage\tCI_lower\tCI_upper\n")
    for line in fin_in:
        fields=line.rstrip().split("\t")
        i_total=int(fields[1])
        i_te=int(fields[2])
        t=proportion_confint(i_te,i_total,alpha=0.05,method='normal')
        fout.write(line.rstrip()+"\t"+str((t[0]+t[1])/2)+"\t"+str(t[0])+"\t"+str(t[1])+"\n")
```
