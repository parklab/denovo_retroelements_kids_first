```python
import numpy as np 
from scipy.stats import t
import statistics

####
def calc_CI(sf_cnt, confidence=0.95):
    l_exon_cnt=[]
    l_high_pLI_cnt=[]
    with open(sf_cnt) as fin:
        for line in fin:
            fields=line.rstrip().split()
            i_exon=float(fields[0])
            i_high_pLI=float(fields[1])
            l_exon_cnt.append(i_exon)
            l_high_pLI_cnt.append(i_high_pLI)

    ci_exon=np.percentile(l_exon_cnt,[100*(1-confidence)/2,100*(1-(1-confidence)/2)]) 
    f_mean_exon=statistics.mean(l_exon_cnt)
    ci_high_pLI=np.percentile(l_high_pLI_cnt,[100*(1-confidence)/2,100*(1-(1-confidence)/2)]) 
    f_mean_pLI=statistics.mean(l_high_pLI_cnt)
    return ci_exon, f_mean_exon, ci_high_pLI, f_mean_pLI

sf_prefix="/Users/simon/research/projects/xTEA_denovo_rslts/rslts_kids_first/summary_results/enrichment_analysis/rslts/simulation_by_cleavage_sits/"

sf_cnt_Alu=sf_prefix+"summarized_cnt_all_rounds_Alu.txt"
ci_exon_alu, f_mean_exon_alu, ci_high_pLI_alu, f_mean_high_pLI_alu = calc_CI(sf_cnt_Alu)

sf_cnt_SVA=sf_prefix+"summarized_cnt_all_rounds_SVA.txt"
ci_exon_sva, f_mean_exon_sva, ci_high_pLI_sva, f_mean_high_pLI_sva = calc_CI(sf_cnt_SVA)

sf_cnt_L1=sf_prefix+"summarized_cnt_all_rounds_L1.txt"
ci_exon_l1, f_mean_exon_l1, ci_high_pLI_l1, f_mean_high_pLI_l1 = calc_CI(sf_cnt_L1)

sf_out_exon=sf_prefix+"exon_95_CI_mean.csv" #
sf_out_pLI=sf_prefix+"high_pLI_CI_mean.csv"
with open(sf_out_exon,"w") as fout_exon, open(sf_out_pLI,"w") as fout_pLI:
    fout_exon.write("Repeat,CI_lower,CI_upper,Mean\n")
    fout_exon.write("Alu,%f,%f,%f\n"%(ci_exon_alu[0], ci_exon_alu[1], f_mean_exon_alu))
    fout_exon.write("L1,%f,%f,%f\n"%(ci_exon_l1[0], ci_exon_l1[1], f_mean_exon_l1))
    fout_exon.write("SVA,%f,%f,%f\n"%(ci_exon_sva[0], ci_exon_sva[1], f_mean_exon_sva))
    
    fout_pLI.write("Repeat,CI_lower,CI_upper,Mean\n")
    fout_pLI.write("Alu,%f,%f,%f\n"%(ci_high_pLI_alu[0], ci_high_pLI_alu[1], f_mean_high_pLI_alu))
    fout_pLI.write("L1,%f,%f,%f\n"%(ci_high_pLI_l1[0], ci_high_pLI_l1[1], f_mean_high_pLI_l1))
    fout_pLI.write("SVA,%f,%f,%f\n"%(ci_high_pLI_sva[0], ci_high_pLI_sva[1], f_mean_high_pLI_sva))

#
#
#####
#####
```
