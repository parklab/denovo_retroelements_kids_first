```python
def gnrt_pwm(sf_seq_freq, i_seq_len, sf_out):
    with open(sf_seq_freq) as fin_seq_freq, open(sf_out,"w") as fout_matrix:
        l_pos_freq=[]
        l_pos_total_cnt=[]##
        for i in range(i_seq_len):
            m_one_pos_freq={'A':0, 'C':0, 'G':0, 'T':0}
            l_pos_freq.append(m_one_pos_freq)
            l_pos_total_cnt.append(0)#
        for line in fin_seq_freq:
            fields=line.rstrip().split()
            s_seq=fields[0]
            i_freq=1
            if len(s_seq)!=i_seq_len:
                continue
            for i in range(i_seq_len):
                l_pos_freq[i][s_seq[i]]+=i_freq
                l_pos_total_cnt[i]+=i_freq
        ####
        for s_char in ['A','C','G','T']:
            s_info=""
            for i in range(i_seq_len):
                s_char_freq=str(float(l_pos_freq[i][s_char])/float(l_pos_total_cnt[i]))
                if s_info=="":
                    s_info=s_char_freq
                else:
                    s_info += (" "+s_char_freq)
            fout_matrix.write(s_info+"\n")

sf_prefix="/Users/simon/research/projects/xTEA_denovo_rslts/rslts_kids_first/summary_results/enrichment_analysis/rslts/simulation_by_cleavage_sits/"
sf_seq=sf_prefix+"simulated_SVA_sites_hg38.gnrted_motifs_all_rounds.txt"
sf_matrix=sf_prefix+"simulated_SVA_sites.matrix"
i_seq_len=7
gnrt_pwm(sf_seq, i_seq_len, sf_matrix)
####
####
```
