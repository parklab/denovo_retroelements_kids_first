```python
sf_info="forweb_cleaned_exac_r03_march16_z_data_pLI_CNV-final.txt"
m_gene_info={}
with open(sf_info) as fin_info, open("1kg_denovo_gene.txt") as fin_genes, open("gene_with_pLI_1kg.tsv","w") as fout:
    for line in fin_info:
        if "transcript" in line:
            continue
        fields=line.rstrip().split()
        s_gene_name=fields[1]
        s_pLI=fields[-4]
        m_gene_info[s_gene_name]=s_pLI
    for line in fin_genes:
        s_gene=line.rstrip()
        if s_gene in m_gene_info:
            fout.write(s_gene+"\t"+m_gene_info[s_gene]+"\n")
            if float(m_gene_info[s_gene])>0.9:
                print(s_gene, m_gene_info[s_gene])
        else:
            fout.write(s_gene+"\tNA\n")
```
